Analysis of the distribution of defaulting clients by CONTRACT_TYPE. Objective: to determine which type of credit agreement posed the highest risk among clients.

```python
data.groupby('NAME_CONTRACT_TYPE')['TARGET'].sum()
```

```python
print(f'{round(data.groupby('NAME_CONTRACT_TYPE')['TARGET'].mean()['Cash loans']*100, 3)}% of bad customers in cash loans')
print(f'{round(data.groupby('NAME_CONTRACT_TYPE')['TARGET'].mean()['Revolving loans']*100, 3)}% of bad customers in revolving loans')
```

```python
data.groupby(['NAME_CONTRACT_TYPE', 'NAME_INCOME_TYPE'])['TARGET'].agg(['count', 'mean'])
```

```python
data_group = data.groupby(['NAME_CONTRACT_TYPE', 'NAME_INCOME_TYPE'])['TARGET'].agg(['count', 'mean'])
data_group['default_rate'] = data_group['mean']*100

income_default_stats = data_group.drop('mean', axis = 1) # дефолтные клиенты по типу доходности
```

```python
income_default_stats
```

Across all sufficiently large NAME_INCOME_TYPE segments, a higher rate of loan default is observed for cash loans than for revolving credit products.

```python
income_default_stats = income_default_stats.reset_index()
```

```python
plot_data_1 = income_default_stats[income_default_stats['count']>100]
plot_data_1
```

```python
plt.figure(figsize=(10, 8))

ax = sns.barplot(data = plot_data_1,
                 x = 'NAME_INCOME_TYPE',
                 y = 'default_rate',
                 hue = 'NAME_CONTRACT_TYPE'
                )   


for container in ax.containers:
    ax.bar_label(container, fmt='%.2f%%')
    
plt.title('Default Rate by Income Type and Contract Type')
plt.xlabel('Income Type')
plt.ylabel('Default Rate (%)')
plt.legend(title='Contract Type')
plt.tight_layout()
plt.show()

# Saving plot

plt.tight_layout()
plt.savefig(
    'outputs/figures/default_rate_by_income_contract.png',
    dpi=300,
    bbox_inches='tight'
)



```

The graph shows that the default rate is higher for Cash loans than for Revolving loans in all major income groups.
